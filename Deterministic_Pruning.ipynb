```python

import math

def confirm_wmoe_ability(phase_angle):
    # 1. Structural Constants
    LAMBDA_THRESHOLD = math.pi * ((1 + 5 ** 0.5) / 2) * (1.0 / 3.0)  # π · φ · (1/3)
    LAMBDA_MAX = 1.10
    GILBERT_BETA = 0.005  # Standard YIG thin-film baseline damping

    print(f"[1] Input Phase Angle: {phase_angle} rad")

    # 2. Phase 1: Injection & Sieve Verification
    # Convert the phase angle token into a continuous physical amplitude
    amplitude = math.sin(phase_angle)
    harmonic_metric = abs(phase_angle * LAMBDA_THRESHOLD)
    print(f"[2] Sieve Ingest -> Amplitude: {amplitude:.4f}, Harmonic Metric: {harmonic_metric:.4f}")

    # 3. Phase 2: Single-Pass Relaxation & Interlock Check
    # Check if the amplitude violates the runaway safety threshold
    if abs(amplitude) > LAMBDA_MAX:
        print("[!] INTERLOCK TRIGGERED: Runaway amplitude. Collapsing to Neutral Ground.")
        return 0  # Immediate hardware phase cancellation

    # Apply exactly one pass of background friction damping
    decayed_amplitude = amplitude * math.exp(-GILBERT_BETA)
    print(f"[3] Pruner Decay -> Settled Amplitude after 1 pass: {decayed_amplitude:.4f}")

    # 4. Phase 3: Ternary Quantization Readout
    DEAD_ZONE = 0.25
    if decayed_amplitude > DEAD_ZONE:
        resolved_state = 1    # Phase-locked Crest
    elif decayed_amplitude < -DEAD_ZONE:
        resolved_state = -1   # Phase-locked Trough
    else:
        resolved_state = 0    # Neutral Ground / Pruned Node

    print(f"[4] Readout Quantization -> Stable Ternary State: {resolved_state}")
    return resolved_state

# =====================================================================
# ABILITY CONFIRMATION EXECUTION
# =====================================================================
print("--- TEST 1: Nominal Crest Optimization ---")
confirm_wmoe_ability(1.2)

print("\n--- TEST 2: Low-Energy Noise Decay (Pruning) ---")
confirm_wmoe_ability(0.1)

print("\n--- TEST 3: Systemic Interlock Over-Excitation Collapse ---")
confirm_wmoe_ability(1.5)  # Generates an ingestion amplitude > 0.99
```
